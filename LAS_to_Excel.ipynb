```python
import lasio
import pandas as pd
```

```python
def las_to_excel(las_file_path, excel_file_path):
    # Load the LAS file
    las = lasio.read(las_file_path)

    # Convert LAS data to a pandas DataFrame
    df = las.df()

    # Create a Pandas Excel writer using openpyxl as the engine.
    writer = pd.ExcelWriter(excel_file_path, engine='openpyxl')

    # Convert the DataFrame to an Excel object.
    df.to_excel(writer, sheet_name='Data')

    # Prepare header information for export
    headers = []
    for section in las.sections.values():
        for item in section:
            headers.append([item.mnemonic, item.unit, item.value, item.descr])
    header_df = pd.DataFrame(headers, columns=['Mnemonic', 'Unit', 'Value', 'Description'])

    # Write the header information in another sheet
    header_df.to_excel(writer, sheet_name='Header')

    # Close the Pandas Excel writer and output the Excel file.
    writer.close()
```

```python
# Example usage
las_to_excel('/content/VU-2.las', 'output_file.xlsx')
```
